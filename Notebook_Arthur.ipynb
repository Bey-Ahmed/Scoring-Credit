Ce notebook présente une analyse de corrélation approfondie du jeu de données `scoring_credit_dataset.xlsx`.

## 1. Chargement des données

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from scipy import stats
from scipy.stats import chi2_contingency, pointbiserialr
import warnings

warnings.filterwarnings('ignore')

# Style global
sns.set_theme(style='whitegrid', palette='muted', font_scale=1.05)
plt.rcParams['figure.dpi'] = 120
plt.rcParams['axes.titlesize'] = 13
plt.rcParams['axes.labelsize'] = 11
```

```python
df = pd.read_excel('scoring_credit_dataset.xlsx', sheet_name='Donnees')
print(f'Dimensions : {df.shape[0]} observations x {df.shape[1]} variables')
df.head(5)
```

```python
# Separation variables numeriques / categorielles / cible
NUM_COLS = [
    'Age', 'Revenu_Annuel', 'Ratio_Endettement_TDSR',
    'Anciennete_Emploi_Ans', 'Anciennete_Residence_Ans',
    'Montant_Demande', 'Duree_Pret_Mois',
    'Score_Credit', 'Nb_Demandes_Recentes'
]

CAT_COLS = [
    'Statut_Emploi', 'Niveau_Education', 'Statut_Residentiel',
    'Etat_Civil', 'Objet_Pret', 'Region', 'Client_Existant'
]

TARGET = 'Defaut_24mois'

print(f'Variables numeriques ({len(NUM_COLS)}) : {NUM_COLS}')
print(f'Variables categorielles ({len(CAT_COLS)}) : {CAT_COLS}')
print(f'Taux de defaut global : {df[TARGET].mean():.1%}')
```

## 2. Matrice de corrélation de Pearson (variables numériques)

La corrélation de Pearson mesure la relation **linéaire** entre deux variables continues.  
Elle varie de -1 (corrélation négative parfaite) à +1 (corrélation positive parfaite).

> **Interprétation usuelle :** |r| < 0.10 → négligeable ; 0.10–0.30 → faible ; 0.30–0.50 → modérée ; > 0.50 → forte.

```python
corr_matrix = df[NUM_COLS].corr(method='pearson')

fig, ax = plt.subplots(figsize=(11, 9))

mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt='.2f',
    cmap='RdBu_r',
    center=0,
    vmin=-1, vmax=1,
    square=True,
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'r de Pearson', 'shrink': 0.8},
    ax=ax
)

ax.set_title('Matrice de correlation de Pearson\n(variables numeriques)', fontweight='bold', pad=15)
ax.set_xticklabels(ax.get_xticklabels(), rotation=35, ha='right')
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Paires les plus correlees (hors diagonale)
corr_pairs = (
    corr_matrix
    .where(np.tril(np.ones(corr_matrix.shape), k=-1).astype(bool))
    .stack()
    .reset_index()
)
corr_pairs.columns = ['Variable 1', 'Variable 2', 'r de Pearson']
corr_pairs['|r|'] = corr_pairs['r de Pearson'].abs()
corr_pairs = corr_pairs.sort_values('|r|', ascending=False).reset_index(drop=True)

print('Top 10 paires les plus correlees (variables numeriques) :')
corr_pairs.head(10)
```

**Observations :**
- Le **Score_Credit** et le **Nb_Demandes_Recentes** présentent une corrélation négative attendue : davantage de demandes récentes signale un risque plus élevé et abaisse le score.
- Le **Ratio_Endettement_TDSR** est positivement corrélé au **Montant_Demande** : un prêt plus important alourdit mécaniquement le ratio.
- La **Anciennete_Emploi_Ans** est positivement corrélée au **Score_Credit** : la stabilité professionnelle améliore l'historique de crédit.
- L'**Age** est positivement corrélé à la **Anciennete_Residence_Ans** : les emprunteurs plus âgés résident en général plus longtemps à la même adresse.

## 3. Corrélation point-bisériale : numériques vs cible (Defaut_24mois)

La corrélation **point-bisériale** (r_pb) est la corrélation de Pearson appliquée entre une variable continue et une variable binaire.  
Elle varie de -1 à +1 et se teste de la même façon (test t).  
Un r_pb positif signifie que la variable tend à être plus élevée chez les défaillants.

```python
pb_results = []
for col in NUM_COLS:
    r, p = pointbiserialr(df[TARGET], df[col])
    pb_results.append({
        'Variable': col,
        'r point-biseriale': r,
        '|r|': abs(r),
        'p-valeur': p,
        'Significatif (alpha=5%)': 'Oui' if p < 0.05 else 'Non'
    })

pb_df = pd.DataFrame(pb_results).sort_values('|r|', ascending=False).reset_index(drop=True)
pb_df
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

colors = ['#d62728' if r > 0 else '#1f77b4' for r in pb_df['r point-biseriale']]
bars = ax.barh(
    pb_df['Variable'],
    pb_df['r point-biseriale'],
    color=colors,
    edgecolor='white',
    height=0.6
)

for i, (_, row) in enumerate(pb_df.iterrows()):
    sig_label = '(**)' if row['Significatif (alpha=5%)'] == 'Oui' else ''
    offset = 0.004 if row['r point-biseriale'] >= 0 else -0.004
    ha = 'left' if row['r point-biseriale'] >= 0 else 'right'
    ax.text(
        row['r point-biseriale'] + offset,
        i,
        f"{row['r point-biseriale']:.3f} {sig_label}",
        va='center', ha=ha, fontsize=9
    )

ax.axvline(0, color='black', linewidth=0.8, linestyle='--')
ax.set_xlabel('r point-biseriale')
ax.set_title('Correlation point-biseriale\n(variables numeriques vs Defaut_24mois)', fontweight='bold')
ax.set_xlim(
    pb_df['r point-biseriale'].min() - 0.1,
    pb_df['r point-biseriale'].max() + 0.12
)

from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#d62728', label='Correlation positive (risque accru)'),
    Patch(facecolor='#1f77b4', label='Correlation negative (risque reduit)')
]
ax.legend(handles=legend_elements, loc='lower right', fontsize=9)

plt.tight_layout()
plt.show()
```

**Observations :**
- Le **Score_Credit** présente la corrélation négative la plus forte : un score élevé est protecteur contre le défaut.
- Le **Ratio_Endettement_TDSR** est la variable numérique la plus corrélée positivement : un endettement élevé augmente le risque.
- Le **Nb_Demandes_Recentes** est positivement associé au défaut : les emprunteurs qui multiplient les demandes sont en général plus fragiles.
- L'**Anciennete_Emploi_Ans** et le **Revenu_Annuel** sont négativement corrélés au défaut : stabilité et revenu réduisent le risque.
- Ces résultats sont cohérents avec les coefficients logistiques du mécanisme génératif.

## 4. V de Cramér : associations entre variables catégorielles

Le **V de Cramér** est une mesure d'association symétrique dérivée du chi-carré de Pearson.  
Il varie de 0 (indépendance totale) à 1 (association parfaite).  

> **Interprétation (Cohen, 1988, ddl=1) :** V < 0.10 → négligeable ; 0.10–0.30 → faible ; 0.30–0.50 → modérée ; > 0.50 → forte.

```python
def cramer_v(x, y):
    contingency = pd.crosstab(x, y)
    chi2, p, dof, _ = chi2_contingency(contingency)
    n = contingency.sum().sum()
    min_dim = min(contingency.shape) - 1
    if min_dim == 0 or n == 0:
        return 0.0, p
    v = np.sqrt(chi2 / (n * min_dim))
    return v, p


n_cat = len(CAT_COLS)
cramer_matrix = pd.DataFrame(np.zeros((n_cat, n_cat)), index=CAT_COLS, columns=CAT_COLS)

for i, col_i in enumerate(CAT_COLS):
    for j, col_j in enumerate(CAT_COLS):
        if i == j:
            cramer_matrix.loc[col_i, col_j] = 1.0
        elif i < j:
            v, _ = cramer_v(df[col_i], df[col_j])
            cramer_matrix.loc[col_i, col_j] = v
            cramer_matrix.loc[col_j, col_i] = v

cramer_matrix = cramer_matrix.astype(float)
```

```python
fig, ax = plt.subplots(figsize=(9, 7))

mask_upper = np.triu(np.ones_like(cramer_matrix, dtype=bool), k=1)

sns.heatmap(
    cramer_matrix,
    mask=mask_upper,
    annot=True,
    fmt='.2f',
    cmap='YlOrRd',
    vmin=0, vmax=1,
    square=True,
    linewidths=0.5,
    linecolor='white',
    cbar_kws={'label': 'V de Cramer', 'shrink': 0.8},
    ax=ax
)

ax.set_title('Matrice de V de Cramer\n(variables categorielles)', fontweight='bold', pad=15)
ax.set_xticklabels(ax.get_xticklabels(), rotation=35, ha='right')
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)
plt.tight_layout()
plt.show()
```

```python
cramer_pairs = []
for i, col_i in enumerate(CAT_COLS):
    for j, col_j in enumerate(CAT_COLS):
        if i < j:
            v, p = cramer_v(df[col_i], df[col_j])
            cramer_pairs.append({
                'Variable 1': col_i,
                'Variable 2': col_j,
                'V de Cramer': round(v, 4),
                'p-valeur': p,
                'Significatif (alpha=5%)': 'Oui' if p < 0.05 else 'Non'
            })

cramer_pairs_df = pd.DataFrame(cramer_pairs).sort_values('V de Cramer', ascending=False).reset_index(drop=True)
print('Associations entre variables categorielles (triees par V de Cramer decroissant) :')
cramer_pairs_df
```

**Observations :**
- Le **Statut_Emploi** est le plus lié au **Statut_Residentiel** : le type de contrat de travail influence le mode d'hébergement.
- Le **Statut_Emploi** est également associé au **Niveau_Education** : le diplôme conditionne en partie le type de contrat.
- L'**Etat_Civil** et le **Statut_Residentiel** sont modérément liés : les personnes mariées tendent davantage à être propriétaires.
- La majorité des paires restent faiblement associées (V < 0.20), limitant les risques de multicolinéarité entre catégorielles.

## 5. Association catégorielles vs cible : V de Cramér et taux de défaut

On mesure l'association entre chaque variable catégorielle et **Defaut_24mois** (variable binaire traitée comme catégorielle à 2 modalités).

```python
cat_target_results = []
for col in CAT_COLS:
    v, p = cramer_v(df[col], df[TARGET])
    cat_target_results.append({
        'Variable': col,
        'V de Cramer': round(v, 4),
        'p-valeur': p,
        'Significatif (alpha=5%)': 'Oui' if p < 0.05 else 'Non'
    })

cat_target_df = pd.DataFrame(cat_target_results).sort_values('V de Cramer', ascending=False).reset_index(drop=True)
cat_target_df
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.barh(
    cat_target_df['Variable'],
    cat_target_df['V de Cramer'],
    color='#2ca02c',
    edgecolor='white',
    height=0.6
)

for i, row in cat_target_df.iterrows():
    sig_label = '(**)' if row['Significatif (alpha=5%)'] == 'Oui' else ''
    ax.text(
        row['V de Cramer'] + 0.003, i,
        f"{row['V de Cramer']:.3f} {sig_label}",
        va='center', fontsize=9
    )

ax.set_xlabel('V de Cramer')
ax.set_xlim(0, cat_target_df['V de Cramer'].max() + 0.08)
ax.set_title('Association categorielles vs Defaut_24mois\n(V de Cramer)', fontweight='bold')
ax.axvline(0.10, color='orange', linestyle='--', linewidth=1, label='Seuil faible (0.10)')
ax.axvline(0.30, color='red', linestyle='--', linewidth=1, label='Seuil modere (0.30)')
ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

```python
n_cols_grid = 2
n_rows_grid = int(np.ceil(len(CAT_COLS) / n_cols_grid))

fig, axes = plt.subplots(n_rows_grid, n_cols_grid, figsize=(14, 4 * n_rows_grid))
axes = axes.flatten()

for idx, col in enumerate(CAT_COLS):
    ax = axes[idx]
    defaut_by_cat = df.groupby(col)[TARGET].mean().sort_values(ascending=False)
    counts = df[col].value_counts()

    bars = ax.bar(
        defaut_by_cat.index,
        defaut_by_cat.values * 100,
        color='#4C72B0', edgecolor='white'
    )

    global_rate = df[TARGET].mean() * 100
    ax.axhline(global_rate, color='red', linestyle='--', linewidth=1.2,
               label=f'Taux global ({global_rate:.1f}%)')

    for bar, mod in zip(bars, defaut_by_cat.index):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.3,
            f'n={counts[mod]}',
            ha='center', va='bottom', fontsize=7.5
        )

    v_val = cat_target_df.loc[cat_target_df['Variable'] == col, 'V de Cramer'].values[0]
    sig = cat_target_df.loc[cat_target_df['Variable'] == col, 'Significatif (alpha=5%)'].values[0]
    sig_label = '(**)' if sig == 'Oui' else ''

    ax.set_title(f'{col}\nV de Cramer = {v_val:.3f} {sig_label}', fontsize=10, fontweight='bold')
    ax.set_ylabel('Taux de defaut (%)')
    ax.set_ylim(0, defaut_by_cat.max() * 100 * 1.25)
    ax.tick_params(axis='x', rotation=25)
    ax.legend(fontsize=8)

for idx in range(len(CAT_COLS), len(axes)):
    axes[idx].set_visible(False)

fig.suptitle('Taux de defaut par modalite – variables categorielles', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

**Observations :**
- Le **Statut_Emploi** est la variable catégorielle la plus discriminante : *SansEmploi* et *Autonome* affichent un taux de défaut supérieur à la moyenne, les *Permanent* sont les moins à risque.
- Le **Client_Existant** montre une association significative : les membres déjà clients ont un taux de défaut inférieur.
- L'**Objet_Pret** différencie les profils : les prêts *Consolidation* sont plus risqués (difficultés préexistantes), *Renovation* et *Auto* sont moins risqués.
- **Region**, **Etat_Civil** et **Niveau_Education** ont des associations plus faibles avec la cible.

## 6. Synthèse globale : ranking de toutes les variables par association avec la cible

```python
summary_num = pb_df[['Variable', '|r|']].rename(columns={'|r|': 'Association avec cible'})
summary_num['Type'] = 'Numerique (|r pb|)'
summary_num['p-valeur'] = pb_df['p-valeur'].values

summary_cat = cat_target_df[['Variable', 'V de Cramer']].rename(columns={'V de Cramer': 'Association avec cible'})
summary_cat['Type'] = 'Categorielle (V Cramer)'
summary_cat['p-valeur'] = cat_target_df['p-valeur'].values

summary_all = pd.concat([summary_num, summary_cat], ignore_index=True)
summary_all = summary_all.sort_values('Association avec cible', ascending=False).reset_index(drop=True)
summary_all['Significatif (alpha=5%)'] = summary_all['p-valeur'].apply(lambda p: 'Oui' if p < 0.05 else 'Non')

summary_all
```

```python
fig, ax = plt.subplots(figsize=(10, 7))

palette = {'Numerique (|r pb|)': '#4C72B0', 'Categorielle (V Cramer)': '#2ca02c'}
bar_colors = [palette[t] for t in summary_all['Type']]

ax.barh(
    summary_all['Variable'],
    summary_all['Association avec cible'],
    color=bar_colors, edgecolor='white', height=0.65
)

for i, row in summary_all.iterrows():
    sig_label = '(**)' if row['Significatif (alpha=5%)'] == 'Oui' else ''
    ax.text(
        row['Association avec cible'] + 0.003, i,
        f"{row['Association avec cible']:.3f} {sig_label}",
        va='center', fontsize=9
    )

ax.set_xlabel("Force d'association avec Defaut_24mois (|r pb| ou V de Cramer)")
ax.set_title('Ranking de toutes les variables\npar association avec la cible (Defaut_24mois)', fontweight='bold', pad=15)
ax.set_xlim(0, summary_all['Association avec cible'].max() + 0.10)

ax.axvline(0.10, color='orange', linestyle='--', linewidth=1, alpha=0.7)
ax.axvline(0.30, color='red', linestyle='--', linewidth=1, alpha=0.7)

from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#4C72B0', label='Numerique (|r point-biseriale|)'),
    Patch(facecolor='#2ca02c', label='Categorielle (V de Cramer)'),
    plt.Line2D([0], [0], color='orange', linestyle='--', label='Seuil faible (0.10)'),
    plt.Line2D([0], [0], color='red', linestyle='--', label='Seuil modere (0.30)'),
]
ax.legend(handles=legend_elements, fontsize=9, loc='lower right')

plt.tight_layout()
plt.show()
```
